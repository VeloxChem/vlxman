### Visualization

The OrbitalViewer provides an interactive interface for visualizing molecular orbitals represented by a VeloxChem molecular orbital object. 

Given a set of molecular orbitals, the viewer constructs three‑dimensional isodensity surfaces corresponding to selected orbitals and enables interactive inspection of their spatial characteristics. The OrbitalViewer operates directly on the orbital coefficients and basis information stored in the VeloxChem molecular orbital data structure, ensuring consistency with the underlying electronic‑structure calculation. 

:::{note}
As an ipywidgets‑based tool, the OrbitalViewer requires a live Jupyter environment and is therefore intended for use in executable notebooks rather than the static HTML version of the manual.
:::

```python
import veloxchem as vlx

molecule = vlx.Molecule.read_name("caffeine")
basis = vlx.MolecularBasis.read(molecule, "6-31g")

scf_drv = vlx.ScfRestrictedDriver()
scf_results = scf_drv.compute(molecule, basis)

orb_viewer = vlx.OrbitalViewer()
orb_viewer.visualize(molecule, basis, scf_drv.mol_orbs)
```

(sec:orbital-localization)=
### Localization

Orbital localization provides an alternative representation of the electronic structure in which molecular orbitals are transformed to maximize their spatial compactness or chemical interpretability, while preserving the span of the underlying orbital subspace. In VeloxChem, localization is performed through unitary rotations of occupied (and, where applicable, virtual) orbitals obtained from a prior electronic structure calculation. The program implements the Boys and Pipek–Mezey localization schemes, which respectively optimize spatial localization and atomic charge separation criteria. Localized orbitals are widely used for qualitative chemical analysis, construction of local correlation methods, and interpretation of bonding patterns in complex molecular systems.

```python
import veloxchem as vlx

molecule = vlx.Molecule.read_name("ethylene")
basis = vlx.MolecularBasis.read(molecule, "cc-pvdz")

scf_drv = vlx.ScfRestrictedDriver()
scf_results = scf_drv.compute(molecule, basis)

loc_drv = vlx.OrbitalLocalizationDriver()

nocc = molecule.number_of_alpha_occupied_orbitals(basis)

boys_results = loc_drv.boys(molecule, basis, scf_results, mo_range=(1, nocc))
pipek_lowdin_results = loc_drv.pipek_mezey(molecule, basis, scf_results, mo_range=(1, nocc), projector="lowdin")
pipek_muliken_results = loc_drv.pipek_mezey(molecule, basis, scf_results, mo_range=(1, nocc), projector="mulliken")
```

The results dictionary of the orbital localization driver contains the localized orbitals in the form of a VeloxChem object that can be passed to the orbital viewer for visual inspection. 

```python
orb_viewer = vlx.OrbitalViewer()
orb_viewer.visualize(molecule, basis, pipek_muliken_results["loc_orbs"])
```
